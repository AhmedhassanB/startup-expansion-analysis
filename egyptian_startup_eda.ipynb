# Dataset Description

## 1. Startups Table

| Column            | Description                                                                    |
| ----------------- | ------------------------------------------------------------------------------ |
| `startup_name`    | Unique name of the startup (**Primary Key**).                                  |
| `industry`        | The sector the startup operates in (e.g., Fintech, E-commerce).                |
| `total_funding_m` | Total funding raised in millions (USD).                                        |
| `status`          | Current operational status (Active, Acquired, Public/Active, Closed, Pivoted). |
| `founder_name`    | Name of the original founder.                                                  |

## 2. Founders Education & Experience Table

| Column              | Description                                                                                          |
| ------------------- | ---------------------------------------------------------------------------------------------------- |
| `startup_name`      | Foreign key linking to the Startups table.                                                           |
| `university_name`   | Bachelor's degree university.                                                                        |
| `university_type`   | University classification (Governmental, Private, or Abroad).                                        |
| `field_of_study`    | Main major or field of study.                                                                        |
| `studied_abroad_yn` | Boolean indicating whether the founder studied abroad (1 = Yes, 0 = No).                             |
| `prev_experience`   | Professional background before founding (Multinational, Entrepreneurial, Corporate).                 |
| `has_mba`           | Boolean indicating whether the founder holds an MBA.                                                 |
| `postgrad_cert_uni` | University or institution of postgraduate studies or high-level certifications (e.g., Harvard, CFA). |
