## Business Questions & Data Analysis

```python
# Which are the top 10 stores generating the highest revenue?
df.groupby('Store ID')['Revenue'].sum().sort_values(ascending=False).head(10)
```

```python
# Which sales region generates the highest total revenue?
df.groupby('Sales Region')['Revenue'].sum().sort_values(ascending=False)
```

```python
# Which region spends more on marketing?
df.groupby('Sales Region')['Marketing Spend'].sum().sort_values(ascending=False)
```

```python
# Which states generate the highest revenue?
df.groupby('State')['Revenue'].sum().sort_values(ascending=False)
```

```python
# How many new and old stores do we have?
df['New Expansion'].value_counts()
```

```python
# Which type of store generates higher total revenue?
df.groupby('New Expansion')['Revenue'].sum().sort_values(ascending=False)
```

```python
# create a profite column 
df['Profit'] = df['Revenue'] - df['Marketing Spend']
df
```
